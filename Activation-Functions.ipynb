## USAGE OF DIFFERENT ACTIVATION FUNCTIONS AND THEIR GRAPHS

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#SIGMOID ACTIVATION FUNCTION
def sigmoid(z):
    sigma = 1 / (1 + np.exp(-z))
    return sigma


#HYPERBOLIC TANGENT(tanh):
def tanh(z):
    e2z = np.exp(2*z)
    return (e2z - 1) / (e2z + 1)

#Rectified Linear Unit
def ReLU(z):
    return np.maximum(0,z)

# Leaky ReLU
def LReLU(z, alpha = 0.1):
    return np.maximum(alpha * z, z)
```

```python
testing_z = np.linspace(-10,10, 100)
testing_z
```

```python
act1 = sigmoid(testing_z)
act2 = tanh(testing_z)
act3 = ReLU(testing_z)
act4 = LReLU(testing_z)
```

```python
fig = plt.figure(figsize = (20,8))
axis = fig.add_subplot(2,4,1)
plt.plot(act1)
plt.grid(True)
axis.set(xlabel='Sigmoid')
axis = fig.add_subplot(2,4,2)
plt.plot(act2)
plt.grid(True)
axis.set(xlabel='Hyperbolic Tangent');
axis = fig.add_subplot(2,4,3)
plt.plot(act3)
plt.grid(True)
axis.set(xlabel='Rectified Linear Unit');
axis = fig.add_subplot(2,4,4)
plt.plot(act4)
plt.grid(True)
axis.set(xlabel='Leaky ReLU');
plt.show()
```
